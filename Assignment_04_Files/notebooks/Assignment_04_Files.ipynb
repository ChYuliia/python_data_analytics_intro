## Assignment 4: DSPAM Confidence Score Processor

## Instructions
Using the file `mbox-short.txt` as the source file:
- Prompt for a filename.
- Open that file and read through it, looking for lines starting with `X-DSPAM-Confidence`.
- Extract the floating-point values from the end of each of those lines.
- Sum the total and count how many lines match.
- Compute the average of those values and produce output like:
  ```text
  Count: 27
  Mean is: 0.75072

````python
from pathlib import Path

# Constant for searching
TARGET_PHRASE = "X-DSPAM-Confidence"

# Prompts for a filename
file_name = input("Enter a file name >> ").strip()
file_path = Path(file_name)

# Check for the existence of the file
if file_path.exists():
    total = 0.0
    count = 0

    # Open the file for reading
    with file_path.open("r", encoding="utf-8") as f:
        for line in f:
            if line.startswith(TARGET_PHRASE):
                # Find the position of the colon
                colon_pos = line.find(":")
                # Extract the part after the colon and convert to float
                value = line[colon_pos + 1 :].strip()
                total += float(value)
                count += 1

    # Output the results
    print(f"Count = {count}")
    if count > 0:
        mean_value = total / count
        print(f"Mean is {mean_value:.5f}")
    else:
        print("Mean is: no data to calculate average")
else:
    print(f"Error: File '{file_name}' not found")
````
