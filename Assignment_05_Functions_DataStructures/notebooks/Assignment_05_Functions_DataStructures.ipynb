## Instructions

### Part 1: Functions
- Write a program to prompt the user for hours and rate per hour using input to compute gross pay. 
- Award time-and-a-half for the hourly rate for all hours worked above 40 hours. 
- Put the logic in a function called `computepay` that returns a value.
- Use 45 hours and a rate of 10.50 per hour to test the program (the pay should be 498.75).

### Part 2: Lists
- Open the file `romeo.txt` and read it line by line. 
- Split each line into a Python list using `.split()`. Build a list of unique words.
- Sort (default sort) and print the resulting words in alphabetical order.

### Part 3: Dictionaries
- Write a program to read through the `mbox-short.txt` file and find who sent the greatest number of mail messages.
- Look for lines starting with `'From '` and take the second word as the sender's email address.
- Create a dictionary mapping the sender's email address to a count, then find the maximum.

## Python Solution

```python
from pathlib import Path
```

```python
# ==========================================
# PART 1: Functions (Gross Pay Calculation)
# ==========================================

def computepay(arg_hours, arg_rate, base_hours=40, ot_rate=1.5):
  if arg_hours < base_hours:
    gross_pay = arg_hours * arg_rate
  else:
    overtime_hours = arg_hours - base_hours
    gross_pay = (base_hours * arg_rate) + (
        overtime_hours * arg_rate * ot_rate
    )
  return gross_pay


try:
  hours_worked = float(input("Enter hours: "))
  hourly_rate = float(input("Enter rate per hour: "))
  print("-" * 22)
  print(f"Gross pay = {computepay(hours_worked, hourly_rate):.2f}")
except ValueError:
  print("Error: Please enter numeric input!")
```

```python
# ==========================================
# PART 2: Lists (Unique Words from romeo.txt)
# ==========================================

FILE_NAME_ROMEO = "romeo.txt"
file_path_romeo = Path(FILE_NAME_ROMEO)
word_list = []

if file_path_romeo.exists():
  with file_path_romeo.open("r", encoding="utf-8") as file_handle:
    for line in file_handle:
      current_line_words = line.split()
      for word in current_line_words:
        if word not in word_list:
          word_list.append(word)

  word_list.sort(key=str.lower)
  output = ", ".join(word_list)
  print(f"List of words ({len(word_list)}):")
  print(f" [{output}]")
else:
  print(f"Error: File '{FILE_NAME_ROMEO}' not found")
```

```python
# ==========================================
# PART 3: Dictionaries (Most Active Sender)
# ==========================================

FILE_NAME_MBOX = "mbox-short.txt"
TARGET_PHRASE = "From "
file_path_mbox = Path(FILE_NAME_MBOX)
counts = {}

if file_path_mbox.exists():
  with file_path_mbox.open("r", encoding="utf-8") as file_handle:
    for line in file_handle:
      if line.startswith(TARGET_PHRASE):
        words = line.split()
        if len(words) > 1:
          email = words[1]
          counts[email] = counts.get(email, 0) + 1

  if counts:
    big_sender = max(counts, key=counts.get)
    big_count = max(counts.values())
    print(f"Most active sender: {big_sender}")
    print(f"Number of messages: {big_count}")
  else:
    print("No message found")
else:
  print(f"Error: File '{FILE_NAME_MBOX}' not found")
```
