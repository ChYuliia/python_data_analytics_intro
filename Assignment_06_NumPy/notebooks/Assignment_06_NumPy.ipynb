```python
import numpy as np

# Optional: make arrays print a bit nicer for class demos
np.set_printoptions(precision=3, suppress=True)

print("NumPy version:", np.__version__)
```

```python
np.random.seed(21)
random_integers = np.random.randint(1,high=500000, size=(20, 5))
random_integers
```

```python
# 1. What is the average value of the second column?
mean_cols = random_integers.mean(axis=0)
print("The average value of the second column:\n", mean_cols[1])

# 2. What is the average value of the first 5 rows of the third and fourth columns?
data_slice = random_integers[:5,2:4]
print("\nThe average value of the first 5 rows of the third and fourth columns:\n", data_slice.mean())

# 3. Create a 10 x 10 array from random_integers array named arrayTenByTen
arrayTenByTen = random_integers.reshape(10, 10).copy()
print("\n10 x 10 array from random_integers array:\n", arrayTenByTen)
```

```python
# 4. Create a new array from the arrayTenByTen that is one rank and called arrayTenFlat
arrayTenFlat = arrayTenByTen.flatten()    # reshape(-1).copy()
print("A new array from the arrayTenByTen that is one rank:\n", arrayTenFlat)

# 5. What is the sum of arrayTenFlat?
print("\nThe sum of arrayTenFlat:\n", arrayTenFlat.sum())
```

```python
# 6. Iterate through arrayTenFlat
print("Iterate through arrayTenFlat:")
for item in np.nditer(arrayTenFlat):
    print(int(item), end=" ")
```

```python
# 7. What is the value of the element (9,2) in random_integers array?
print("The value of the element (9,2) in random_integers array:\n", random_integers[9,2])

# 8. What is the data type of arrayTenFlat?
print("\nThe data type of arrayTenFlat:\n", arrayTenFlat.dtype)

# 9. In arrayTenFlat replace the value in index 5 with 42
arrayTenFlat[5] = 42
print("\nIn arrayTenFlat replace the value in index 5 with 42:\n", arrayTenFlat)
```

```python
# 10. Save the array random_integers to a file.  List the directory showing the saved array.
np.save("week6_homework_array.npy", random_integers)
np.savetxt("week6_homework_array.csv", random_integers, delimiter=",", fmt="%.2f")

print("Saved: week6_homework_array.npy and week6_homework_array.csv\n")
!dir week6_homework_array*
```
