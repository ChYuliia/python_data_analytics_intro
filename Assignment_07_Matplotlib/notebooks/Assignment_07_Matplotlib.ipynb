```python
# Run this to get started (no output unless error)
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

# Line plots

### Generate a lineplot of y1 and y2 on x axis


```python
# Using x, y1 and y2 
# Generate of y1 and y2 on x axis
x = np.linspace(0, 20)
y1 = np.sin(x)
y2 = np.sin(x - np.pi)

#Code goes below:
plt.plot(x, y1)
plt.plot(x, y2)
```

## Color and style

### Create same plot above with:
### - y1 color is black, linestyle solid, linewidth of 2, squar marker at a size of 6
### - y2 color is gray, linestyle dashed, linewidth of 2, triangle markers at size of 6



```python
# Using:
x = np.linspace(0, 20)
y1 = np.sin(x)
y2 = np.sin(x - np.pi)

# Create same plot above with:
# - y1 color is black, linestyle solid, linewidth of 2, squar marker at a size of 6
# - y2 color is gray, linestyle dashed, linewidth of 2, triangle markers at size of 6

# Code goes below
fig, ax = plt.subplots()
ax.plot(x, y1, color="black", linestyle="solid", linewidth=2, marker="s", markersize=6)
ax.plot(x, y2, color="gray", linestyle="dashed", linewidth=2, marker='^', markersize=6)

ax.set_title("Line Plot")

plt.show()
```

# Vertical bar charts

### Create a vertical bar chart for the values by years. 
### Its ok to have years split i.e, (2012.0, 2012.5...)


```python
# Using
years = np.arange(2012, 2015)
values = [2, 5, 9]
# Create a vertical bar chart values by years.

# Code goes below:
fig, ax = plt.subplots()
ax.bar(years, values)

ax.set_title("Bar Chart: Values by Year")
ax.set_xlabel("Year")
ax.set_ylabel("Values")
ax.grid(True, axis='y', alpha=0.3)

plt.show()
```

## Create a multiseries bar chart with:
* category1_values: Color = Blue, No edgecolor, width = 0.4, align = center and a label of y1
* category2_values: Color = Orange, No edgecolor, width = 0.4, align = center and a label of y2


```python
# Using:
years = np.arange(2012, 2015)
category1_values = [2, 5, 9]
category2_values = [7, 6, 3]

# Create a multiserise bar chart with:
#   category1_values: Color = Blue, No edgecolor, width = 0.4, align = center and a label of y1
#   category2_values: Color = Orange, No edgecolor, width = 0.4, align = center and a label of y2

# Code goes below:
x = np.arange(len(years))
width = 0.4 

fig, ax = plt.subplots()
ax.bar(x - width/2, category1_values, width, label="y1", color="blue", 
       edgecolor="none", align="center")
ax.bar(x + width/2, category2_values, width, label="y2", color="orange", 
       edgecolor="none", align="center")

ax.set_xticks(x, years)
ax.set_title("Multiseries Bar Chart")
ax.set_xlabel("Year")
ax.set_ylabel("Values")
ax.legend()
ax.grid(True, axis='y', alpha=0.3)

plt.show()
```

# Horizontal bar charts
### Create a horizontal bar chart for the given values:
* categories = ['A', 'B', 'C', 'D', 'E']
* values = [7, 12, 4, 2, 9]
* Lable each as the letter from the categories parameter... i.e.  A, B, C ...


```python
# Using
categories = ['A', 'B', 'C', 'D', 'E']
values = [7, 12, 4, 2, 9]

# Create a horizontal bar chart for the values
# Lable each as "With the letter in the categories variable... i.e.  A,  B, C, D ...

# Code goes below:
fig, ax = plt.subplots()

ax.barh(categories, values)

ax.set_title("Horizonatal Bar: Values by Category")
ax.set_xlabel("Values")
ax.set_ylabel("Categories")
ax.grid(True, axis='x', alpha=0.3)

plt.show()
```

### Create a horizontal bar chart for the given values with:
* Color = Blue, no edgecolor, height of 0.6, center aligned


```python
# Using:
categories = ['A', 'B', 'C', 'D', 'E']
values = [7, 12, 4, 2, 9]

# Create a horizontal bar chart with
# Color = Blue, no edgecolor, height of 0.6, center aligned

# Code goes below:
fig, ax = plt.subplots()

ax.barh(categories, values, color="blue", edgecolor="none", height=0.6, align="center")

ax.set_title("Horizontal Bar Chart")
ax.set_xlabel("Values")
ax.set_ylabel("Categories")
ax.grid(True, axis='x', alpha=0.3)
plt.show()
```

# Pie charts

## Create a pie chart
* For the given counts = [17, 14]


```python
# Using
counts = [17, 14]

# Create a pie chart

# Code goes below:
fig, ax = plt.subplots()
ax.pie(counts)
ax.set_title("Pie Chart")
plt.show()
```

### Create a pie chart for counts = [17, 14] using:
* Colors of blue and orange
* Labels of 'Category A' and 'Other categories'
* Set the angle at 90
* Set the percentage display format of 1.1



```python
# Using
counts = [17, 14]

# Create a pie chart for counts = [17, 14] using:
#  Colors of blue and orange
#  Labels of 'Category A' and 'Other categories'
#  Set the angle at 90
#  Set the percentage display format of 1.1

# Code goes below:
labels = ['Category A', 'Other categories']
fig, ax = plt.subplots()
ax.axis('equal')

ax.pie(counts, labels=labels, autopct='%1.1f%%', colors=['blue', 'orange'], startangle=90)

ax.set_title("Pie Chart: Category Distribution")
plt.show()
```

# Scatter plots

## Create a scatter plot for the given data



```python
# Using
x = range(20)
y = np.arange(50, 70) + (np.random.random(20) * 10.)

# Create a scatter plot for the given data

#Code goes below:
fig, ax = plt.subplots()

ax.scatter(x, y)

ax.set_title("Scatter Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)

plt.show()
```

## Create a scatter plot with the given data with:
* Color of red
* Marker 

* `c`: Set the color of the markers.

* `s`: Set the size of the markers.

* `marker`: Set the marker style, e.g., circles, triangles, or squares.

* `edgecolor`: Set the color of the lines on the edges of the markers.

```python
x = np.arange(20)
y = np.arange(50, 70) + (np.random.random(20) * 10.)

#Code goes below:
fig, ax = plt.subplots()

ax.scatter(x, y, c="red", s=100, marker='^', edgecolor="black")

ax.set_title("Scatter Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)

plt.show()
```
