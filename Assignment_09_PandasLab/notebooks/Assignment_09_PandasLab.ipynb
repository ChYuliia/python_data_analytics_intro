````python
# Be sure we import MapPlotLib as plt and NumPy as np to be used with Pandas.
import matplotlib.pyplot as plt
````

**16.** Basic plotting: plot
### (2 points) Create a basic Bar Plot with the following data frame 'df'



````python
#Use:
df = pd.DataFrame(np.random.rand(10,4),columns=['a','b','c','d'])

#Code goes below:
ax = df.plot.bar(rot=0, width=0.8, figsize=(10, 6))

plt.title('Random Data Bar Plot')
plt.xlabel('Original Row Index')
plt.ylabel('Values')

plt.show()
````

**17.** Basic Plotting: Stacked Plots


#### (2 points) produce a stacked bar plot using the data frame 'df'

````python
#Use:
df = pd.DataFrame(np.random.rand(10,4),columns=['a','b','c','d'])

#Code goes below:
ax = df.plot.bar(stacked=True, rot=0, width=0.8, figsize=(10, 6))

plt.title('Random Stacked Bar Plot')
plt.xlabel('Original Row Index')
plt.ylabel('Values')

plt.show()
````

**18.** Plotting: Histograms

#### (2 points) Create a histogram with 20 bins using the data frame ‘df’

````python
#Use:
df = pd.DataFrame({'a':np.random.randn(1000)+1,'b':np.random.randn(1000),'c':
np.random.randn(1000) - 1}, columns=['a', 'b', 'c'])

#Code goes below:
ax = df.plot.hist(bins=20, figsize=(10, 5), edgecolor='white')

plt.title('Frequency Distribution of Variables a, b, and c')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
````

**19.** Plotting: Scatter plots
#### (2 points) Create a scatter plot of columns `b` and `d` using the data frame ‘df’

````python
# Use
df = pd.DataFrame(np.random.rand(50, 4), columns=['a', 'b', 'c', 'd'])

#Code goes below:
ax = df.plot.scatter(x='b', y='d', c='DarkBlue', s=50, figsize=(8, 6))

plt.title('Relationship Between Columns b and d')
plt.xlabel('b')
plt.ylabel('d')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

**20.** Plotting: Box Plots
