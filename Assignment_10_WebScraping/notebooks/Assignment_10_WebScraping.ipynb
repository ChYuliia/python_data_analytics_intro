## Web Scraping

**Instruction:**

1. Scrape all http URL links from the Beautiful Soup Main Page:
2. https://www.crummy.com/software/BeautifulSoup/
3. The final output should be a list of all URLs from the page.

```python
# Import required libraries
import requests
from bs4 import BeautifulSoup

# URL of the webpage
url = "https://www.crummy.com/software/BeautifulSoup/"
headers = {"User-Agent": "Mozilla/5.0"}

try:
    # Send request to website
    response = requests.get(url, headers=headers, timeout=10)
    response.raise_for_status()
    html = response.text
    print(">> Live request worked")

    # Parse HTML
    soup = BeautifulSoup(html, "html.parser")

    # Create a list to store links
    links = []

    # Find all <a> tags with href attribute
    for a in soup.find_all("a", href=True):
        href = a["href"]
        if href.startswith("http:"):
            links.append(href)
        
    # Print all extracted links
    print(f">> Links found: {len(links)}")
    for link in links:
        print(link)
        
except Exception as e:
    print(">> Live request failed")
    print(">> Reason:", e) 
```
