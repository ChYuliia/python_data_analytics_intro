Using the iris dataset create a summary data plot for the following:
* Sepal length by Sepal width by species
* Petal length by Petal width by species 

### **1. Load and Explore Data**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
filename = "iris.csv"
iris_df = pd.read_csv(filename)

# Display basic data overview
print("Summary statistics:")
display(iris_df.describe(include='all'))

# Check for data integrity (missing values)
print("Missing values by column:")
print(iris_df.isna().sum())

# Verify class distribution
print("\nSpecies distribution:")
iris_df["species"].value_counts()
```

### **2. Sepal Analysis (Manual Plotting)**

```python
plt.figure(figsize=(10, 6))

colors_dict = {'setosa': '#0077B6', 'versicolor': '#FF8920', 
               'virginica': '#1F9A1F'}

# Plotting each species manually
for species, color in colors_dict.items():
    subset = iris_df[iris_df["species"] == species]
    plt.scatter(subset["sepal_length"], subset["sepal_width"], 
                c=color, label=species, linewidth=1.5, 
                marker = 'o', alpha=0.7)

# Formatting the plot
plt.title("Iris Analysis: Sepal Length vs. Sepal Width by Species", 
          fontsize=16, fontweight='bold')
plt.xlabel("Sepal Length", fontsize=12)
plt.ylabel("Sepal Width", fontsize=12)
plt.grid(True, linestyle=':', alpha = 0.6)
plt.legend(title='Species', fontsize=10)
plt.show()
```

### **3. Petal Analysis (Using Seaborn)**

```python
plt.figure(figsize=(10, 6))
# Using Seaborn for more efficient categorical plotting
sns.scatterplot(data=iris_df, x='petal_length', y='petal_width', 
                palette=colors_dict, hue='species', s=75, 
                alpha=0.7, linewidth=1.0)

# Formatting the plot
plt.title("Iris Analysis: Petal Length vs. Petal Width by Species", 
          fontsize=16, fontweight='bold')
plt.xlabel("Petal Length", fontsize=12)
plt.ylabel("Petal Width", fontsize=12)
plt.grid(True, linestyle=':', alpha = 0.6)
plt.show()
```

### **4. Visualizing all feature combinations to identify key differentiators (Pairplot)**

```python
sns.pairplot(iris_df, hue='species', palette=colors_dict, 
             markers=['o', 's', 'D'], plot_kws={'s': 20}) 
plt.suptitle("Pairwise Relationships of Iris Features", 
             y=1.02, fontsize=16)
plt.show()
```

### **5. Feature Correlation: Identifying mathematical relationships between physical attributes**

```python
plt.figure(figsize=(10,8))
corr = iris_df.drop('species', axis=1).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', 
            linewidths=1, square=True, cbar_kws={"shrink": .8})
plt.title("Correlation Matrix of Iris Attributes")
plt.show()
```
